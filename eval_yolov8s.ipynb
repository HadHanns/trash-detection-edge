# YOLOv8s Evaluation Notebook
Notebook ini digunakan untuk memuat model YOLOv8s, menjalankan evaluasi pada dataset validasi, dan menampilkan hasilnya (Confusion Matrix, kurva, dll) secara interaktif.

```python
from ultralytics import YOLO
import os
from IPython.display import Image, display

# Memuat model terbaik
model_path = 'weights/best_yolov8sclass.pt'
model = YOLO(model_path)
print(f'Model berhasil dimuat dari: {model_path}')
```

```python
# Menjalankan evaluasi pada validation set
# Plot evaluasi akan disimpan otomatis
results = model.val(
    data='data/datasets/RFT.yaml',
    imgsz=640,
    batch=4,
    device=0,
    plots=True,
    project='C:/yolo_out',
    name='eval_notebook',
    exist_ok=True,
    verbose=True
)
```

```python
# 1. Menampilkan Confusion Matrix (Absolute)
conf_matrix_path = 'C:/yolo_out/eval_notebook/confusion_matrix.png'
if os.path.exists(conf_matrix_path):
    display(Image(filename=conf_matrix_path, width=800))
else:
    print('Confusion matrix tidak ditemukan.')
```

```python
# 2. Menampilkan Confusion Matrix (Normalized)
conf_matrix_norm_path = 'C:/yolo_out/eval_notebook/confusion_matrix_normalized.png'
if os.path.exists(conf_matrix_norm_path):
    display(Image(filename=conf_matrix_norm_path, width=800))
else:
    print('Normalized Confusion matrix tidak ditemukan.')
```
